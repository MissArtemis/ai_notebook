```python
class Fenwick:
    def __init__(self, n: int):
        self.n = n
        self.tree = [0] * (n + 1)

    def add(self, i: int, delta: int) -> None:
        """第 i 个位置增加 delta，下标从 1 开始"""
        while i <= self.n:
            self.tree[i] += delta
            i += i & -i

    def prefix_sum(self, i: int) -> int:
        """查询前 i 个位置的和"""
        total = 0
        while i > 0:
            total += self.tree[i]
            i -= i & -i
        return total

    def range_sum(self, l: int, r: int) -> int:
        """查询闭区间 [l, r] 的和"""
        return self.prefix_sum(r) - self.prefix_sum(l - 1)
    

class Solution:
    def minOperations(self, nums: list[int], locked: list[int], k: int) -> int:
        max_num = 0
        lock_cnt = 0
        n = len(nums)
        for i, num in enumerate(nums):
            max_num = max(max_num, num)
            if locked[i]:
                if num !=i+1 or max_num > i+1:
                    lock_cnt += 1
                    if lock_cnt > k:
                        return -1
        bit = Fenwick(n)
        swap_cnt = 0

        for i, num in enumerate(nums):
            swap_cnt += i - bit.prefix_sum(num)
            bit.add(num, 1)
        return swap_cnt+lock_cnt

        

            
```

```python
nums = [2,1,3,4]
locked = [0,1,0,1]
k = 1
solution = Solution()
ans = solution.minOperations(nums, locked, k)
print(ans)
```
