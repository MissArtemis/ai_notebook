```python
# 1. Solution 1 BFS
from collections import deque


class Solution:
    def isUgly(self, n: int) -> bool:

        queue = deque([])
        queue.append(1)
        visited = set()
        visited.add(1)

        while queue:
            size = len(queue)
            for i in range(size):
                cur = queue.popleft()
                if cur == n:
                    return True
                elif cur >n:
                    break
                else:
                   if cur * 2 <= n and cur * 2 not in visited:
                    queue.append(cur*2)
                    visited.add(cur*2) 
                   if cur * 3 <= n and cur * 3 not in visited:
                    queue.append(cur*3)
                    visited.add(cur*3) 
                   if cur * 5 <= n and cur * 5 not in visited:
                    queue.append(cur*5) 
                    visited.add(cur*5)
        
        return False

        
```

```python
# 2. Solution 2 
class Solution:
    def isUgly(self, n: int) -> bool:
        if n <= 0:
            return False

        for factor in (2, 3, 5):
            while n % factor == 0:
                n //= factor

        return n == 1
```

```python
n = 6
solution = Solution()
print(solution.isUgly(n))
```
