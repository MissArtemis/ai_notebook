```python
from math import sqrt
class Solution:
    def countPrimes(self, n: int) -> int:

        def is_prime(n):
            max_factor = int(sqrt(n))
            for i in range(2,max_factor+1):
                if n%i ==0:
                    return False
            return True
        cnt = 0
        for i in range(2,n):
            if is_prime(i):
                cnt+=1

        return cnt
```

```python
n = 10
solution = Solution().countPrimes(n)
print(solution)
```
