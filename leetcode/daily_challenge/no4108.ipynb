## DFS：全蓝费用减去最大节省

```python
# size 是完整子树大小；saving 是包含当前节点的红色连通块最大收益。
# 红点加 size、蓝点减 size，只接入正收益孩子；答案为全蓝费用减去全局最大收益。
```

```python
from typing import List
class Solution:
    def minCost(self, n: int, colors: str, edges: List[List[int]]) -> int:
        g = [[] for _ in range(n)]
        for u, v in edges:
            g[u].append(v)
            g[v].append(u)
        cost_all_blue = 0
        max_saving = 0

        def dfs(u, parent):
            nonlocal cost_all_blue, max_saving
            size = 1
            saving = 0
            for v in g[u]:
                if v == parent:
                    continue
                child_size, child_saving = dfs(v, u)
                size += child_size
                saving += max(0, child_saving)

            if colors[u] == 'R':
                cost_all_blue += size
                saving += size
            else:
                saving -= size
            max_saving = max(max_saving, saving)
            return size, saving
        dfs(0, -1)
        return cost_all_blue - max_saving
```

```python
n = 18
colors = "RBBBRRRRRRRBRRRRRR"
edges = [
    [0, 1], [0, 2], [0, 3],
    [1, 4], [1, 5],
    [4, 6], [6, 7],
    [5, 8], [8, 9],
    [2, 10], [2, 11], [11, 12],
    [3, 13], [13, 14], [14, 15], [15, 16], [16, 17]
]
solution = Solution()
result = solution.minCost(n, colors, edges)
print(result)
```
