```python
class Solution:
    def countValidSubstrings(self, s: str) -> int:
        ans = 0
        nearest_a = -1
        nearest_ab = -1
        nearest_abc= -1
        for i, ch in enumerate(s):
            if ch == "a":
                nearest_a = i
            elif ch == "b":
                nearest_ab = nearest_a
            else:
                nearest_abc = nearest_ab
            ans += i - nearest_abc
        return ans
            
            
```

```python
s = "abac"
solution = Solution()
result = solution.countValidSubstrings(s)
print(result)
```
