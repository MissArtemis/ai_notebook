```python
from functools import lru_cache


class Solution:
    def minimumCost(self, cost1: int, cost2: int, costBoth: int, need1: int, need2: int) -> int:
        @lru_cache(None)
        def dfs(n1,n2, cost):
            nonlocal need1,need2,cost1,cost2,costBoth
            if n1>=need1 and n2>=need2:
                return cost
            else:
                if n1>=need1:
                    return min(dfs(n1,n2+1,cost+cost2),dfs(n1+1,n2+1,cost+costBoth))
                elif n2>=need2:
                    return min(dfs(n1+1,n2,cost+cost1),dfs(n1+1,n2+1,cost+costBoth))
                else:
                    return min(dfs(n1+1,n2,cost+cost1),dfs(n1,n2+1,cost+cost2),dfs(n1+1,n2+1,cost+costBoth)) 
        return dfs(0,0,0)
```

```python
class Solution:
    def minimumCost(
        self, cost1: int, cost2: int, costBoth: int,
        need1: int, need2: int
    ) -> int:
        shared = min(need1, need2)

        return (
            shared * min(costBoth, cost1 + cost2)
            + (need1 - shared) * min(cost1, costBoth)
            + (need2 - shared) * min(cost2, costBoth)
        )
```

```python
cost1 = 5
cost2 = 4
costBoth = 15
need1 = 2
need2 = 3
solution = Solution()
result = solution.minimumCost(cost1,cost2,costBoth,need1,need2)
print(result)
```
