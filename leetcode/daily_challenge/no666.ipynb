## 提交版

````python
# 字典键是深度和层内位置，孩子位置为 2 * idx - 1、2 * idx。
# total 传入时不含当前节点；真实叶子结算路径和，空节点返回 0。
````

````python
from typing import List


class Solution:
    def pathSum(self, nums: List[int]) -> int:
        value_record = {num // 10: num % 10 for num in nums}

        def has_child(depth, idx):
            return any(
                (depth + 1) * 10 + child_idx in value_record
                for child_idx in (2 * idx - 1, 2 * idx)
            )

        def dfs(depth, idx, total):
            key = depth * 10 + idx
            if key not in value_record:
                return 0

            total += value_record[key]
            if not has_child(depth, idx):
                return total

            return (
                dfs(depth + 1, 2 * idx - 1, total)
                + dfs(depth + 1, 2 * idx, total)
            )

        return dfs(1, 1, 0)
````
