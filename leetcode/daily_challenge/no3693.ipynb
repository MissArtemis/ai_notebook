# 3693. 爬楼梯 II：向前更新 DP

[原文](https://leetcode.cn/problems/climbing-stairs-ii/solution/3693-pa-lou-ti-iilou-ti-kai-shi-shou-fei-32gg/)

```python
# dp[i] 表示到达第 i 级台阶的最小费用；从当前台阶向后更新 1、2、3 级。
# 每次支付落点费用与步长平方，costs 前补 0 对齐台阶编号。
```

```python
from typing import List

class Solution:
    def climbStairs(self, n: int, costs: List[int]) -> int:
        dp = [float('inf')] * (n+1)
        dp[0] = 0
        costs = [0] + costs
        for i in range(n+1):
            if i+1<n+1:
                dp[i+1] = min(dp[i+1],costs[i+1]+1+dp[i])
            if i+2<n+1:
                dp[i+2] = min(dp[i+2], costs[i+2]+4+dp[i])
            if i+3<n+1:
                dp[i+3] = min(dp[i+3], costs[i+3]+9+dp[i])
        # print(dp)
        return dp[-1]
```

```python
n = 4
costs = [5,1,6,2]
solution = Solution()
result = solution.climbStairs(n, costs)
print(result)
```
