```python
from typing import List
class Solution:
    def minSelection(self, availableIntervals: List[List[int]], targetIntervals: List[List[int]]) -> int:
        sorted_target = sorted(targetIntervals, key=lambda x: x[0])
        merged_target = []
        ans = 0
        prev = targetIntervals[0] if targetIntervals else None
        merged_target.append(prev)
        for i in range(1, len(sorted_target)):
            prev = merged_target.pop(-1)
            prev_start = prev[0]
            prev_end = prev[1]
            curr_start = sorted_target[i][0]
            curr_end = sorted_target[i][1]
            if curr_start <= prev_end:
                prev[1] = max(prev_end, curr_end)
                merged_target.append(prev)
            else:
                merged_target.append(prev)
                merged_target.append(sorted_target[i])
        merged_target = [interval for interval in merged_target if interval]
        print(merged_target)
        sorted_available = sorted(availableIntervals, key=lambda x: x[0])
        prev_available = sorted_available[0] if sorted_available else None
        start = prev_available[0] if prev_available else None
        end = prev_available[1] if prev_available else None
        for i in range(1, len(sorted_available)):
            if sorted_available[i][0] == start:
                end = sorted_available[i][1]
                if start <= merged_target[0][0] and end >= merged_target[0][1]:
                    ans += 1
                    merged_target.pop(0)
            elif sorted_available[i][0] < end:
                end = max(end, sorted_available[i][1])
                if start <= merged_target[0][0] and end >= merged_target[0][1]:
                    ans += 1
                    merged_target.pop(0)
            else:
                start = sorted_available[i][0]
                end = sorted_available[i][1]
        if merged_target:
            return -1

        return ans 
```

```python
availableIntervals = [[1,4],[1,7],[3,9],[5,10],[10,15]]
targetIntervals = [[1,3],[2,5],[7,8],[10,13],[11,15],[12,13]]
solution = Solution()
result = solution.minSelection(availableIntervals, targetIntervals)
print(result)
```
