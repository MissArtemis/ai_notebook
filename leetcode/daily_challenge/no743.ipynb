# 743. 网络延迟时间：Dijkstra 模板

[原文](https://leetcode.cn/problems/network-delay-time/solution/743-wang-luo-yan-chi-shi-jian-yong-yi-da-23q4/)

```python
# 按有向边建图，Dijkstra 求起点 k 到各节点的最短距离，跳过堆里的过期记录。
# 答案取 dist[1:] 的最大值；有节点不可达就返回 -1。
```

```python
#Dijkstra Template

from heapq import heappop, heappush
from math import inf


def dijkstra(g, start):
    dist = [inf] * len(g)
    dist[start] = 0

    # (从起点出发的累计距离, 节点)
    heap = [(0, start)]

    while heap:
        d, u = heappop(heap)

        # 当前记录已经过期
        if d > dist[u]:
            continue

        for v, w in g[u]:
            new_dist = d + w

            # 找到更短路线：更新距离，并加入候选记录
            if new_dist < dist[v]:
                dist[v] = new_dist
                heappush(heap, (new_dist, v))

    return dist

class Solution:
    def networkDelayTime(
        self, times: list[list[int]], n: int, k: int
    ) -> int:
        # 第一步：把题目输入转换成邻接表
        g = [[] for _ in range(n + 1)]

        for u, v, w in times:
            g[u].append((v, w))

        # 第二步：调用模板，求从 k 到所有节点的最短时间
        dist = dijkstra(g, k)

        # 第三步：根据题目要求，从距离数组中提取答案
        ans = max(dist[1:])

        return -1 if ans == inf else int(ans)
```

## 原文样例

```python
times = [[2,1,1],[2,3,1],[3,4,1]]
n = 4
k = 2
solution = Solution()
ans = solution.networkDelayTime(times,n,k)
print(ans)
```
