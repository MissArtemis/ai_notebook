```python
# Solution 1
#  prefix-sum
class Solution:
    def findMaxConsecutiveOnes(self, nums: list[int]) -> int:
        ans = 0
        prefix = [0]
        prev_sum = 0
        for i in range(len(nums)):
            if i>= 1 and nums[i] == 0 and nums[i-1] == 0:
                prev_sum = 0
                prefix.append(0)
            elif i>= 1 and nums[i] == 0 and nums[i-1]!=0:
                prev_sum = prefix[-1]
                prefix.append(0)
                
            else:
                prefix.append(prefix[-1] + nums[i])
            ans = max(ans,prev_sum + prefix[-1]+1)
        return min(ans, len(nums))
```

```python
#Solution 2 DP
class Solution:
    def findMaxConsecutiveOnes(self, nums: list[int]) -> int:
        dp = [[0] * 2 for _ in range(len(nums))]
        dp[0][0] = nums[0]
        dp[0][1] = 1
        for i in range(1, len(nums)):
            if nums[i] == 1:
                dp[i][0]=dp[i-1][0] + 1
                dp[i][1] = dp[i-1][1] + 1
            else:
                dp[i][0] = 0
                dp[i][1] = dp[i-1][0] + 1
        # print(dp)
        return max(row[1] for row in dp)
        

        
```

```python
#Solution 3 space-optimized dynamic programming
class Solution:
    def findMaxConsecutiveOnes(self, nums: list[int]) -> int:
        dp0 = dp1 = 0
        ans = 0
        for num in nums:
            if num == 1:
                dp0 +=1
                dp1 +=1
            else:
                dp1 = dp0+1
                dp0 = 0
            ans = max(ans,dp0,dp1)
        return ans
```

```python
nums = [1, 0, 1, 1, 0, 1]
solution = Solution().findMaxConsecutiveOnes(nums)
print(solution)
```
