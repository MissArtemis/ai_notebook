# 1021. 删除最外层的括号

删除每个原语的最外层括号。两种解法的时间复杂度都是 O(n)。

## Solution 1：记录断点，再切片

`cnt` 回到 0 时，记录当前右括号的位置。第一段内部是 `s[1:close]`；后续段内部是 `s[prev_close + 2:close]`。

踩坑：后续段从 `prev_close + 1` 开始会保留下一段的外层左括号。反例 `()()` 的错误结果是 `(`，正确结果是空字符串。

```python
class Solution:
    def removeOuterParentheses(self, s: str) -> str:
        break_points = [0]
        cnt = 0
        for i, c in enumerate(s):
            if c == "(":
                cnt += 1
            elif c == ")":
                cnt -= 1
                if cnt == 0:
                    break_points.append(i)
            else:
                continue

        return "".join(
            s[1:break_points[1]] if i == 1
            else s[break_points[i - 1] + 2:break_points[i]]
            for i in range(1, len(break_points))
        )
```

## Solution 2：按深度一趟扫描

左括号：先判断 `depth > 0` 是否保留，再增加深度。右括号：先减少深度，再判断是否保留。

外层左括号出现前深度为 0，外层右括号处理后深度为 0，所以二者都不会进入结果。

时间 O(n)，Python 结果列表与返回字符串占 O(n) 空间。深度变量占 O(1) 空间。省掉断点和切片步骤，但两种解法的时间复杂度相同，实际速度取决于实现与输入。

在力扣单独提交第二种解法时，把 `Solution2` 改为 `Solution`。

```python
class Solution2:
    def removeOuterParentheses(self, s: str) -> str:
        result = []
        depth = 0

        for c in s:
            if c == "(":
                if depth > 0:
                    result.append(c)
                depth += 1
            else:  # 题目保证 s 只含合法括号
                depth -= 1
                if depth > 0:
                    result.append(c)

        return "".join(result)
```

## 样例对照

结果应与预期答案 `expected` 比较。`ans == s` 只是在比较结果和原输入，不能判断解法是否正确。

```python
s = "(()())(())(()(()))"
expected = "()()()()(())"
for solution in (Solution(), Solution2()):
    ans = solution.removeOuterParentheses(s)
    print(type(solution).__name__, repr(ans), ans == expected)
```
