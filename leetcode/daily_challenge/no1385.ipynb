# 1385. 两个数组间的距离值：暴力、二分与双指针

## 暴力

````python
# 对每个 a1 检查 arr2；一旦发现距离不超过 d 的元素，就停止并排除当前 a1。
````

````python
# Solution 1

from typing import List, Optional, Tuple, Dict, Set
class Solution:
    def findTheDistanceValue(self, arr1: List[int], arr2: List[int], d: int) -> int:
        ans = 0
        for a1 in arr1:
            flag = 0
            for a2 in arr2:
                if abs(a1 - a2) <=d:
                    flag = 1
                    break
            if not flag:
                ans+=1
        return ans
                    
````

## 二分区间左端点

````python
# 定位第一个不小于 a1-d 的元素；它不存在或大于 a1+d 时，闭区间内没有元素。
````

````python
# Solution 2 bisect

from typing import List, Optional, Tuple, Dict, Set
from bisect import bisect_left, bisect_right
class Solution:
    def findTheDistanceValue(self, arr1: List[int], arr2: List[int], d: int) -> int:
        ans = 0
        arr2 = sorted(arr2)
        for a1 in arr1:
            i = bisect_left(arr2,a1-d)
            if i == len(arr2) or arr2[i]>a1+d:
                ans+=1

        return ans
````

## 二分左右邻居

````python
# 只检查 a1 左右最近的两个邻居，距离必须都严格大于 d。
# List 沿用前面的类型导入；此版会原地排序 arr2。
````

````python
# Solution 3 bisect
from bisect import bisect_left, bisect_right
class Solution:
    def findTheDistanceValue(self, arr1: List[int], arr2: List[int], d: int) -> int:
        ans = 0
        arr2.sort()
        for a1 in arr1:
            i = bisect_left(arr2,a1)
            if i == len(arr2) or abs(a1-arr2[i])>d:
                if i == 0 or abs(a1-arr2[i-1])>d:
                    ans+=1
        return ans
````

## 双指针

````python
# 两个数组排序后，j 始终指向 arr2 中第一个不小于 a-d 的位置，只向右移动。
# 此版会原地排序两个数组；同一个 arr2 元素可以影响多个 a。
````

````python
#Solution 4 double pointer
from typing import List, Optional, Tuple, Dict, Set
from bisect import bisect_left, bisect_right
class Solution:
    def findTheDistanceValue(self, arr1: List[int], arr2: List[int], d: int) -> int:
        arr1.sort()
        arr2.sort()
        ans = 0
        j = 0
        for a in arr1:
            while j < len(arr2) and arr2[j]< a-d:
                j+=1
            if j== len(arr2) or arr2[j]>a+d:
                ans+=1
        return ans
````

## 原文样例

顺序运行时调用上面的双指针版本，输出为 `2`。

````python
arr1 = [4,5,8]
arr2 = [10,9,1,8]
d = 2
solution = Solution()
ans = solution.findTheDistanceValue(arr1,arr2,d)
print(ans)
````
