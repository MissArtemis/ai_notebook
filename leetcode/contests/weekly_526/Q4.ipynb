```python
class Solution:
    def minActivationCost(self, edges: list[list[int]], values: list[int]) -> int:
        n = len(values)
        graph = [[] for _ in range(n)]
        for u, v in edges:
            graph[u].append(v)
            graph[v].append(u)
        bit_cnt = [x.bit_count()%2 for x in values]
```
