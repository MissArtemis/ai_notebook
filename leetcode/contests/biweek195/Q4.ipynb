```python
class Solution:
    def repeatedStringMatch(self, a: str, b: str, x: int, y: int) -> bool:
        length_a, length_b = len(a), len(b)
        total_a, total_b = len(a) * x, len(b) * y
        if total_b > total_a:
            return False
        min_times = min(y, length_a // length_b + 2)
        pattern = b * min_times
        text_a_seq = a * (len(pattern) // length_a + 2)
        start = text_a_seq.find(pattern)
        if start == -1:
            return False
        if start + length_b * y > length_a * x:
            return False

        return True
    
```

```python
a = "ab"
b = "abab"
x = 2
y = 1
solution = Solution()
result = solution.repeatedStringMatch(a, b, x, y)
print(result)
```
