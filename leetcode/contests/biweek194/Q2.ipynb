```python
class Solution:
    def countSubarrays(self, nums: list[int], minK: int, maxK: int) -> int:
        ans = 0
        cnt = 0
        for idx, num in enumerate(nums):
            if num < minK or num > maxK:
                cnt = 0
            else:

                cnt+=1
            ans += cnt
        return ans
```

```python
nums = [1,4,7,3]
minK = 1
maxK = 5
solution = Solution()
print(solution.countSubarrays(nums, minK, maxK))
```
