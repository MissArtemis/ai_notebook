# 3644. 排序排列：错位元素按位与

[原文](https://leetcode.cn/problems/maximum-k-to-sort-a-permutation/solution/3644ti-mu-rang-wo-pai-xu-wo-que-zhi-xie-6sxl7/)

```python
# 只对 i != x 的元素取按位与；结果对应数组中可作交换中转的数字。
# 0 错位时直接返回 0；全部有序时把初始值 -1 转成 0。
```

```python
from typing import List


class Solution:
    def sortPermutation(self, nums: List[int]) -> int:
        if nums[0]: 
            return 0
        ans = -1 
        for i, x in enumerate(nums):
            if i != x:
                ans &= x
        return max(ans, 0)
```
