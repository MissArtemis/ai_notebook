## Solution 1：错误版本（仅供复盘）

只保留当前最小异或值会丢失后续最优路径。反例 [[0, 1, 0], [2, 0, 3]] 的正确答案为 1，这版返回 2。

```python
'''
Solution 1.
Minimum XOR Path in a Grid
首先想到，因为题目要求规定从起点出发(0,0)，到固定的终点结束(m-1,n-1)
因此，其实对于任意一个坐标(i,j)，存在客观上的最小cost值。
所以想到用动态规划，用dp[i][j]来记录每条路径上到坐标(i,j)的值，
取最小的作为dp[i][j]最后的值
最后需要的答案即是dp[m-1][n-1]
'''

class Solution:
    def minCost(self, grid: list[list[int]]) -> int:
        m,n = len(grid),len(grid[0])
        dp = [[float('inf')] * n for _ in range(m)]
        dp[0][0] = grid[0][0]
        for i in range(m):
            for j in range(n):
                cur_value = grid[i][j]
                if i > 0:
                    dp[i][j] = min(dp[i][j], dp[i-1][j] ^ cur_value)
                if j > 0:
                    dp[i][j] = min(dp[i][j], dp[i][j-1] ^ cur_value)
        print(dp)
        return dp[-1][-1]
        
```

## Solution 2：二维集合 DP

````python
# 每格保存所有可达的累计异或值，相同结果合并；从上方和左方转移。
````

````python
#Solution 2
class Solution:
    def minCost(self, grid: list[list[int]]) -> int:
        m,n = len(grid),len(grid[0])
        dp = [[set() for _ in range(n)] for _ in range(m)]
        dp[0][0].add(grid[0][0])
        for i in range(m):
            for j in range(n):
                if i > 0:
                    for val in dp[i-1][j]:
                        dp[i][j].add(val ^ grid[i][j])
                if j > 0:
                    for val in dp[i][j-1]:
                        dp[i][j].add(val ^ grid[i][j])
        # print(dp)
        return min(dp[-1][-1])
````

## Solution 3：滚动数组

````python
# 从左到右更新：dp[j] 是上方旧集合，dp[j - 1] 是本行左方集合。
# 先生成新集合再覆盖，避免遍历时修改原集合。
````

````python
# Solution 3：滚动数组，只保留一行的异或值集合
class Solution:
    def minCost(self, grid: list[list[int]]) -> int:
        m, n = len(grid), len(grid[0])
        dp = [set() for _ in range(n)]
        dp[0].add(grid[0][0])

        for i in range(m):
            for j in range(n):
                if i == 0 and j == 0:
                    continue

                cur_value = grid[i][j]
                # 更新前，dp[j] 是上方格子的集合。
                # 新建集合，避免在遍历旧集合时修改它。
                current = {val ^ cur_value for val in dp[j]}
                if j > 0:
                    # 从左到右更新，所以 dp[j - 1] 已是本行左方格子的集合。
                    current.update(val ^ cur_value for val in dp[j - 1])
                dp[j] = current

        return min(dp[-1])

# 每个数最多有 10 个二进制位，累计异或值也只可能为 0～1023。
# 时间复杂度：O(m * n * 1024)，与 Solution 2 相同。
# 空间复杂度：O(n * 1024)，由二维数组缩减为一行。
````
