```python
class Solution:
    def countStaircase(self, nums: list[int]) -> int:
        n = len(nums)
        ans = 0
        for left in range(n):
            target_cnt = {}
            cur_cnt = {}


            for right in range(left, n):
                cur = nums[right]
                print("target", target_cnt)
                print("cur", cur_cnt)
                if cur not in target_cnt:
                    target_cnt[cur] = len(target_cnt) + 1
                    cur_cnt[cur]=0
                cur_cnt[cur]+=1
                if cur_cnt == target_cnt:
                    ans += 1
                elif cur_cnt[cur] > target_cnt[cur]:
                    break
                else:
                    continue
                
        return ans
```

```python
nums =  [2,3,4,4,3,4]
solution = Solution()
print(solution.countStaircase(nums))
```
