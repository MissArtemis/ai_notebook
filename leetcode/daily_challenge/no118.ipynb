```python
class Solution:
    def generate(self, numRows: int) -> list[list[int]]:
        prev = [1]
        ans = []
        ans.append(prev)
        for row in  range(2,numRows+1):
            seq = [0] * row
            seq[0],seq[-1] = 1,1
            length = len(seq)
            for l in range(1,length-1):
                seq[l] = prev[l-1] + prev[l]
            ans.append(seq)
            prev = seq

        return ans
```

```python
numRows = 5
solution = Solution()
print(solution.generate(numRows))
```
