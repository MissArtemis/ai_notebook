```python
#BFS
from math import floor
from math import sqrt

class Solution:
    def numSquares(self, n: int) -> int:
        queue = []
        queue.append(n)
        visited = set()
        visited.add(n)
        step = 0

        while queue:
            size = len(queue)
            for i in range(size):
                cur = queue.pop(0)
                sq = floor(sqrt(cur))
                for r in range(sq,-1,-1):
                    if cur >= r**2:
                        if cur - r**2 in visited:
                            continue
                        if cur == r**2:
                            return step+1
                        else:
                            queue.append(cur - r**2)
                            visited.add(cur - r**2)
                    else:
                        break
            step +=1
                

        return step

n = 13
ans = Solution().numSquares(n)
ans
```

```python
#DFS

from functools import lru_cache
from math import sqrt
from math import floor



class Solution:
    
    def numSquares(self, n: int) -> int:
        @lru_cache(None)
        def backtrack(cur):
            if cur == 0:
                return 0

            ans = float('inf')
            sq_root = floor(sqrt(cur))

            for r in range(sq_root, 0, -1):
                if cur - r**2 >= 0:
                    ans = min(ans, backtrack(cur - r**2) + 1)
                else:
                    break

            return ans

        return backtrack(n)


n = 13
ans = Solution().numSquares(n)
ans
```

```python
#DP
from collections import defaultdict
from functools import lru_cache
from math import sqrt
from math import floor

class Solution:
    def numSquares(self, n: int) -> int:
        record = defaultdict(lambda:float('inf'))

        def backtrack(cur):
            nonlocal record

            if cur == 0:
                return 0

            ans = float('inf')
            sq_root = floor(sqrt(cur))

            for r in range(sq_root, 0, -1):
                next_cur = cur - r**2

                if record[next_cur] == float('inf'):
                    ans = min(ans, backtrack(next_cur) + 1)
                else:
                    ans = min(ans, record[next_cur] + 1)

            record[cur] = ans
            return ans
        

        return backtrack(n)

n = 13
ans = Solution().numSquares(n)
ans
```
