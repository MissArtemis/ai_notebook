```python
class Solution:
    def smallestRepunitDivByK(self, k: int) -> int:
        candidate_record ={
            "1":1,
            "11":2,
            "111":3,
            "1111":4,
            "11111":5,
            "111111":6,
            # "1111111":7,
            # "11111111":8,
            # "111111111":9
        }
        for c in candidate_record.keys():
            if c%k==0:
                return candidate_record[c]

        
        return -1
```

```python
k = 3
solution = Solution()
ans = solution.smallestRepunitDivByK(k)
print(ans)
```
