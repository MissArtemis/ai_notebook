## 双堆调度

````python
# 空闲堆按 (权重, 下标) 选服务器，忙碌堆按 (完成时间, 下标) 释放。
# 分配前释放所有已完成服务器；没有空闲服务器时，时间跳到最早完成时刻并再次全部释放。
````

````python
from typing import List
import heapq
class Solution:
    def assignTasks(self, servers: List[int], tasks: List[int]) -> List[int]:
        idle_servers = [(w,idx) for idx,w in enumerate(servers)]
        heapq.heapify(idle_servers)
        busy_servers = []

        ans = []
        ts = 0
        def release():
            while busy_servers and busy_servers[0][0]<=ts:
                end_time, idx = heapq.heappop(busy_servers)
                heapq.heappush(idle_servers,(servers[idx],idx))

        for idx,task in enumerate(tasks):
            ts = max(idx,ts)
            release()
            if not idle_servers:
                ts = busy_servers[0][0]
                release()

            weight, idx = heapq.heappop(idle_servers)
            ans.append(idx)
            heapq.heappush(busy_servers,(ts+task,idx))


        return ans
````

````python
servers = [3,3,2]
tasks = [1,2,3,2,1,2]
solution = Solution()
result = solution.assignTasks(servers,tasks)
print(result)
````

## heapq 基本操作

````python
import heapq

heap = [5, 2, 8]
heapq.heapify(heap)

print(heap[0])                # 2，只看，不删除
heapq.heappush(heap, 1)       # 加入 1
print(heapq.heappop(heap))    # 1，取走最小值
print(heapq.heappop(heap))    # 2，现在的最小值
````

## 堆排序模板

````python
import heapq


def heap_sort(nums):
    # 复制一份，避免 heapify 和后面的弹出操作修改输入。
    heap = list(nums)
    heapq.heapify(heap)

    ordered = []
    while heap:
        ordered.append(heapq.heappop(heap))

    return ordered


print(heap_sort([5, 2, 8, 1, 2]))  # [1, 2, 2, 5, 8]
````

## 元组优先级

````python
import heapq

idle = [(3, 0), (3, 1), (2, 2)]
heapq.heapify(idle)

print(heapq.heappop(idle))  # (2, 2)：权重更小
print(heapq.heappop(idle))  # (3, 0)：权重相同，下标更小
print(heapq.heappop(idle))  # (3, 1)
````

## 双堆调度伪代码

```plainText
建立空闲堆：按资源的选择优先级排序
建立忙碌堆：按资源恢复可用的时间排序

按照题目规定的顺序遍历任务：
    当前时间 = max(当前时间, 任务到达时间)
    释放所有恢复时间 <= 当前时间的资源

    如果没有空闲资源：
        当前时间 = 最早的资源恢复时间
        再次释放所有恢复时间 <= 当前时间的资源

    从空闲堆取最优资源
    记录分配结果
    计算资源恢复时间，放入忙碌堆
```
