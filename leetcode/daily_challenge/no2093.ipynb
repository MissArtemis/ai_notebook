## 普通 Dijkstra 对照模板

```python
#Dijkstra Template

from heapq import heappop, heappush
from math import inf


def dijkstra(g, start):
    dist = [inf] * len(g)
    dist[start] = 0

    # (从起点出发的累计距离, 节点)
    heap = [(0, start)]

    while heap:
        d, u = heappop(heap)

        # 当前记录已经过期
        if d > dist[u]:
            continue

        for v, w in g[u]:
            new_dist = d + w

            # 找到更短路线：更新距离，并加入候选记录
            if new_dist < dist[v]:
                dist[v] = new_dist
                heappush(heap, (new_dist, v))

    return dist
```

## 分层 Dijkstra

```python
# dist[u][used] 表示恰好使用 used 次折扣到达 u 的最小费用；每条边分别尝试原价与半价。
# 公路双向，折扣不必用完；答案取终点各层的最小值。
```

```python
from heapq import heappop, heappush
from math import inf


def dijkstra(g, start, discounts):
    # dist[u][used]：
    # 到达 u，恰好使用 used 次折扣的最小费用
    dist = [[inf] * (discounts + 1) for _ in g]
    dist[start][0] = 0

    # (累计费用, 城市, 已使用折扣次数)
    heap = [(0, start, 0)]

    while heap:
        d, u, used = heappop(heap)

        # 跳过这个状态的过期记录
        if d > dist[u][used]:
            continue

        for v, w in g[u]:
            # 选择一：不使用折扣
            new_dist = d + w

            if new_dist < dist[v][used]:
                dist[v][used] = new_dist
                heappush(heap, (new_dist, v, used))

            # 选择二：使用一次折扣
            if used < discounts:
                new_dist = d + w // 2

                if new_dist < dist[v][used + 1]:
                    dist[v][used + 1] = new_dist
                    heappush(heap, (new_dist, v, used + 1))

    return dist


class Solution:
    def minimumCost(
        self,
        n: int,
        highways: list[list[int]],
        discounts: int,
    ) -> int:
        # 公路是双向的，城市编号为 0 ~ n-1
        g = [[] for _ in range(n)]

        for u, v, w in highways:
            g[u].append((v, w))
            g[v].append((u, w))

        dist = dijkstra(g, 0, discounts)

        # 不要求用完折扣，取到达终点的所有合法状态的最小值
        ans = min(dist[n - 1])

        return -1 if ans == inf else int(ans)
```

## 原文样例

```python
n = 5
highways = [[0,1,4],[2,1,3],[1,4,11],[3,2,3],[3,4,2]]
discounts = 1
ans = Solution().minimumCost(n,highways,discounts)
print(ans)
```
