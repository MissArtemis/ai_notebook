```python
class Solution:
    def maximizeSum(self, nums: list[int], k: int) -> int:
        increment = []
        for idx, num in enumerate(nums):
            max_num = int(''.join(sorted(str(num), reverse=True)))
            increment.append(max_num - num)
        increment.sort(reverse=True)
        return sum(nums) + sum(increment[:k])
```

```python
 nums = [102,90,14]
k = 2
solution = Solution()
print(solution.maximizeSum(nums, k))
```
