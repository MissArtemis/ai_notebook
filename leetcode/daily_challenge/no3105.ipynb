```python
from typing import List
class Solution:
    def longestMonotonicSubarray(self, nums: List[int]) -> int:
        if len(nums) == 1:
            return 1
        increse_anchor = nums[0]
        decrease_anchor = nums[0]
        increase_score = 1
        decrease_score = 1
        max_decrease_score = 1
        max_increase_score = 1
        for i in range(1, len(nums)):
            increse_anchor = decrease_anchor = nums[i]
            if nums[i] > nums[i-1]:
                increase_score+=1
                max_increase_score = max(max_increase_score,increase_score)
                decrease_score=1
            elif nums[i] < nums[i-1]:
                
                increase_score = 1
                decrease_score +=1
                max_decrease_score = max(decrease_score,max_decrease_score)
            else:
                increase_score = 1
                decrease_score = 1

                

            

        return max(max_decrease_score,max_increase_score)
```

```python
nums = [3, 2, 1]
solution = Solution()
result = solution.longestMonotonicSubarray(nums)
print(result)
```
