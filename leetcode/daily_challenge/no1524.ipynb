```python
from typing import List

class Solution:
    def numOfSubarrays(self, arr: list[int]) -> int:
        MOD = 10**9 + 7
        total = 0
        odd, even = 0, 1
        ans = 0

        for a in arr:
            total +=a
            ans += (odd if total%2==0 else even)
            if total%2 == 0:
                even+=1
            else:
                odd+=1
        return ans % MOD
```

```python
arr = [1, 3, 5]
solution = Solution().numOfSubarrays(arr)
print(solution)
```
