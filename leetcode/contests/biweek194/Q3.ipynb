```python

from collections import defaultdict


class Solution:
    def countMagicalPairs(self, s: str) -> int:
        ans = 0
        n = len(s)

        for j in range(n-1):

            left_set = set()
            left_record = [0] * 27
            for i in range(j, -1, -1):
                left_set.add(s[i])
                left_record[len(left_set)] += 1
            right_set = set()
            right_record = [0] * 27
            for k in range(j+1, n):
                right_set.add(s[k])
                right_record[len(right_set)] += 1
            for k in range(1, 27):
                ans += left_record[k] * right_record[k]

        return ans
```

```python
s = "abcac"
ans = Solution().countMagicalPairs(s)
print(ans)
```
