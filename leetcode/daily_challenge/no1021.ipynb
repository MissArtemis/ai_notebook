# 1021. 删除最外层的括号

[原文](https://leetcode.cn/problems/remove-outermost-parentheses/solution/1021-shan-chu-zui-wai-ceng-de-gua-hao-sh-0p55/)

## 深度扫描

````python
# 左括号先判断再加深度，右括号先减深度再判断。
````

````python
class Solution:
    def removeOuterParentheses(self, s: str) -> str:
        result = []
        depth = 0

        for c in s:
            if c == "(":
                if depth > 0:
                    result.append(c)
                depth += 1
            else:
                depth -= 1
                if depth > 0:
                    result.append(c)

        return "".join(result)
````

## 断点切片

````python
# 后续原语的内部从上一个右括号下标加 2 开始。
````

````python
class Solution:
    def removeOuterParentheses(self, s: str) -> str:
        break_points = [0]
        cnt = 0

        for i, c in enumerate(s):
            if c == "(":
                cnt += 1
            else:
                cnt -= 1
                if cnt == 0:
                    break_points.append(i)

        return "".join(
            s[1:break_points[1]] if i == 1
            else s[break_points[i - 1] + 2:break_points[i]]
            for i in range(1, len(break_points))
        )
````

## 错误切片片段

````python
# 下方 +1 会保留下一段的外层左括号。
````

## 原文样例

````python
s = "(()())(())(()(()))"
expected = "()()()()(())"
ans = Solution().removeOuterParentheses(s)
print(ans == expected)
````
