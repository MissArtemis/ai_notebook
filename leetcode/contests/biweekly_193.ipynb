## 1. 和为目标值的最大乘积对

[比赛提交](https://leetcode.cn/submissions/detail/753965686/)

```python
# 哈希表查找补数，只比较 num>diff 的配对；最大乘积从负无穷开始。
```

```python
class Solution:
    def maxProductPair(self, nums: list[int], target: int) -> list[int]:
        max_product = -float("inf")
        record = {num: i for i, num in enumerate(nums)}
        result = [-1, -1]

        for i, num in enumerate(nums):
            diff = target - num
            if diff not in record or num <= diff:
                continue

            product = num * diff
            if product > max_product:
                max_product = product
                result = [i, record[diff]]

        return result
```

## 2. 最长有韧性的子数组 I

[比赛提交](https://leetcode.cn/submissions/detail/753973599/)

```python
# 扫描连续同余段，缩短长度直到 (长度-1)*余数 能被 k 整除。
```

```python
class Solution:
    def resilientSubarray(self, nums: list[int], k: int) -> int:

        left = 0
        size = len(nums)
        ans = 1
        while left < size:
            remainder = nums[left] % k
            right = left + 1
            while right <size and nums[right] % k == remainder:
                right += 1
            length = right - left
            while ((length-1)*remainder) % k != 0:
                length -= 1

            ans = max(ans, length)
            left = right
        return ans
```
