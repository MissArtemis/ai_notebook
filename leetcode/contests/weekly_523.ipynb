## 1. 连续三项斐波那契数之和

[比赛提交](https://leetcode.cn/submissions/detail/754006172/)

```python
# 滚动维护连续三项，三项和超过 n 后停止。
```

```python
class Solution:
    def threeFibonacciSum(self, n: int) -> bool:
        a = 0
        b = 1
        c = 1
        while a + b + c <= n:
            if a + b + c == n:
                return True
            else:
                a , b, c = b, c, b + c
        return False
```

## 2. 选择质数子集 I

[比赛提交](https://leetcode.cn/submissions/detail/754008365/)

```python
# 筛出质数后从小到大选择，优先保证数量最多。
```

```python
class Solution:
    def maxPrimes(self, n: int, s: int) -> list[int]:
        if n < 2:
            return []
        prime_num_flag = [True] * (n + 1)
        prime_num_flag[0] = prime_num_flag[1] = False
        for i in range(2, int(n ** 0.5) + 1):
            if prime_num_flag[i]:
                for j in range(i * i, n + 1, i):
                    prime_num_flag[j] = False
        primes = [i for i in range(2, n + 1) if prime_num_flag[i]]
        ans = []
        for prime in primes:
            if s >= prime:
                ans.append(prime)
                s -= prime
            if s == 0:
                break
        return ans
```

## 3. 设计公交车座位预订系统

[比赛提交](https://leetcode.cn/submissions/detail/754014755/)

```python
# 记录每次预订的实际费用，取消时原样扣回。
# 最短时间等于当前座位数，即总费用减去每笔 3 秒预订多付的 2 秒。
```

```python
class BusBooking:

    def __init__(self, n: int):
        self.record = {}
        self.save_time_cnt = 0
        self.total_time = 0
        
        

    def toggle(self, seat: str) -> None:
        seat_letter = seat[0]
        seat_number = seat[1:]
        if seat in self.record:
            if self.record[seat] == 3:
                self.save_time_cnt-=1
            self.total_time -= self.record.pop(seat)
        else:
            if seat_letter == 'A':
                if 'B' + seat_number not in self.record:
                    self.total_time += 1
                    self.record[seat] = 1
                else:
                    self.total_time += 3
                    self.save_time_cnt+=1
                    self.record[seat] = 3
            elif seat_letter == 'D':
                if 'C' + seat_number not in self.record:
                    self.total_time += 1
                    self.record[seat] = 1
                else:
                    self.total_time += 3
                    self.save_time_cnt+=1
                    self.record[seat] = 3
            else:
                self.total_time += 1
                self.record[seat] = 1
                
        

    def getTotalTime(self) -> int:
        return self.total_time
        

    def getMinTime(self) -> int:
        return self.total_time - 2 * self.save_time_cnt
        


# Your BusBooking object will be instantiated and called as such:
# obj = BusBooking(n)
# obj.toggle(seat)
# param_2 = obj.getTotalTime()
# param_3 = obj.getMinTime()
```

## 4. 选择质数子集 II

[比赛提交](https://leetcode.cn/submissions/detail/754021923/)

```python
# 倒序更新 0/1 背包；同总和下先取数量少的，再取字典序小的。
```

```python
class Solution:
    def maxPrimeSubset(self, n: int, s: int) -> list[int]:
        if n < 2:
            return []
        prime_num_flag = [True] * (n + 1)
        prime_num_flag[0] = prime_num_flag[1] = False
        for i in range(2, int(n ** 0.5) + 1):
            if prime_num_flag[i]:
                for j in range(i * i, n + 1, i):
                    prime_num_flag[j] = False
        primes = [i for i in range(2, n + 1) if prime_num_flag[i]]
        ans = []
        dp = [None] * (s + 1)
        dp[0] = []
        max_sum = 0
        for prime in primes:
            for i in range(s, prime - 1, -1):
                if dp[i-prime] is None:
                    continue
                subset = dp[i-prime]
                cur = subset + [prime]
                if (dp[i] is None) or (len(cur)<len(dp[i])) or (len(cur)==len(dp[i]) and cur<dp[i]) :
                    dp[i] = cur
                    max_sum = max(max_sum, i)

        # print(candidates)
        return dp[max_sum] if max_sum > 0 else []
```
