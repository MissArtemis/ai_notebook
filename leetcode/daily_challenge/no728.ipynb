```python
class Solution:
    def selfDividingNumbers(self, left: int, right: int) -> list[int]:

        def valid(num:int):
            digit_nums = [int(s) for s in str(num)]
            print(digit_nums)
            for d in digit_nums:
                if d == 0:
                    return False
                if num%d!=0:
                    return False
            return True
        candidates = range(left,right+1)
        ans = [c for c in candidates if valid(c)]
        return ans
            
```

```python
left = 1
right = 22
solution = Solution()
ans = solution.selfDividingNumbers(left,right)
print(ans)
```
