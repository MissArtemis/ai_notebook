```python
class Solution:
    def countValidSubstrings(self, s: str) -> int:
        ans = 0
        for i in range(len(s)):
            for j in range(i + 1, len(s) + 1):
                substring = s[i:j]
                if "abc" in substring:
                    break
                else:
                    ans += 1
        return ans
```

```python
# Solution 2
class Solution:
    def countValidSubstrings(self, s: str) -> int:
        ans = 0
        start = -1
        for i in range(len(s)):
            if i >= 2 and s[i-2:i+1] == "abc":
                start = i - 2
            ans += i - start
        return ans
```

```python
s = "abac"
solution = Solution()
result = solution.countValidSubstrings(s)
print(result)
```
