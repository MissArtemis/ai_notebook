```python
'''
优先考虑dp场景
对于dp[n][k]，表示把长度为1到n的木棍排成一列，从左边恰好能看到k根的方案数
能看到的木棍一定从左到右越来越长，但不一定是最长的k根
例如[1,3,2]能看到1和3，而不是最长的2和3
那么对于dp[n][k]，它的前置状态其实就两个，dp[n-1][k]和dp[n-1][k-1]
具体看最后一根木棍是不是这n根中最长的，这里讨论的是最终排列的末尾，不要求按长度从小到大添加木棍
1.如果是，那么最后一根一定能被看到，也不会影响前面木棍的可见性
去掉它，前面n-1根恰好能看到k-1根，这一类共有dp[n-1][k-1]种方案
2.如果不是，由于最长的木棍一定在它前面，所以最后一根一定看不到
从n-1根非最长的木棍中选任意一根放在最后，一共有n-1种选择
选定后，剩余n-1根仍然需要恰好能看到k根；具体长度不影响计数，只取决于相对大小
因此可以按长度从小到大重新编号为1到n-1，每种选择对应dp[n-1][k]种排列
不同选择的最后一根不同，不会重复计数，这一类共有(n-1) * dp[n-1][k]种方案

综上，状态转移方程为 dp[n][k] = dp[n-1][k-1] + (n-1) * dp[n-1][k]
显然dp[1][1] = 1，为了遍历方便，计算一下dp[0][0]
dp[1][1] = dp[0][0] + 0*dp[0][1] = dp[0][0] = 1
其余状态先初始化为0；i>0时dp[i][0]=0，因为至少能看到第一根木棍
所以j从1开始遍历，且只需要计算j<=i的状态
每次转移时取模，避免中间整数过大；时间复杂度O(n*k)，空间复杂度O(n*k)
'''
class Solution:
    def rearrangeSticks(self, n: int, k: int) -> int:
        mod = 10**9 + 7
        dp = [[0]* (k+1) for _ in range(n+1)]
        dp[0][0] = 1
        for i in range(1,n+1):
            for j in range(1,min(i,k)+1):
                dp[i][j] = (dp[i-1][j-1] + (i-1)*dp[i-1][j]) % mod

        return dp[-1][-1]
```

```python
n = 20
k = 11
solution = Solution()
print(solution.rearrangeSticks(n,k))
```
