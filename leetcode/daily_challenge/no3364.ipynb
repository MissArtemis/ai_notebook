```python
from typing import List

class Solution:
    def minimumSumSubarray(self, nums: List[int], l: int, r: int) -> int:
        ans = float('inf')
        for i in range(len(nums)):
            for j in range(i+l,i+r+1):
                # print(j)
                if j <= len(nums):
                    cost = sum(nums[i:j])
                    # print(nums[i:j])
                    # print(cost)
                    if cost>0:
                        ans = min(ans,cost)
                else:
                    break
        return -1 if ans == float('inf') else ans
```

```python
nums = [3, -2, 1, 4]
l = 2
r = 3
solution = Solution()
result = solution.minimumSumSubarray(nums,l,r)
print(result)
```
