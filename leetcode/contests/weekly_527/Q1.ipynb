```python
class Solution:
    def findMismatchedSegments(self, s1: str, s2: str) -> list[list[str]]:
        ans = []
        n = len(s1)
        i = 0
        while i<n:
            if s1[i] == s2[i]:
                i+=1
                continue
            start = i
            while i < n:
                if s1[i]!=s2[i]:
                    i+=1
                else:
                    break
            ans.append([s1[start:i],s2[start:i]])
        return ans
```

```python
s1 = "earth"
s2 = "empty"
solution = Solution()
result = solution.findMismatchedSegments(s1, s2)
print(result)
```
