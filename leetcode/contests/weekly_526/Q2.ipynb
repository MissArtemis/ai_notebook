```python
from math import gcd
class Solution:
    def minimumMaxValue(self, n: int, index: int, value: int) -> int:
        if n == 1:
            return value
        if value == 1:
            return -1
        if value % 2 == 0:
            p = 2
        else:
            p = 3
            while p * p <= value and value % p != 0:
                p += 2
            if p * p > value:
                p = value
        right_cnt = n - index - 1
        left_cnt = index
        ans = value + p + 2*(right_cnt - 1)

        left_min_value = value - p - 2*(left_cnt - 1)
        if left_min_value <= 0:
            return -1

        
        
        return ans
```

```python
n = 4
index = 2
value = 8
solution = Solution()
a = solution.minimumMaxValue(n, index, value)
print(a)
```
