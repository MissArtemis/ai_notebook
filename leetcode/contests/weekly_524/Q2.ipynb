```python
class Solution:
    def lastVisitedCell(self, grid: list[list[str]]) -> list[int]:
        record = {
            "L": (0, -1),
            "R": (0, 1),
            "U": (-1, 0),
            "D": (1, 0)
        }
        visited = set()
        m, n = len(grid), len(grid[0])
        visited.add((0, 0))
        i, j = 0, 0
        while 0 <= i < m and 0 <= j < n:
            di, dj = record[grid[i][j]]
            ni, nj = i + di, j + dj
            if not (0 <= ni < m and 0 <= nj < n) or (ni, nj) in visited:
                break
            visited.add((ni, nj))
            i, j = ni, nj

        return [i, j]
```

```python
grid = [["R","D"],["U","L"]]
solution = Solution()
result = solution.lastVisitedCell(grid)
print(result)
```
