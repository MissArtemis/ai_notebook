```python
from collections import Counter
class Solution:
    def countSuffix(self, nums: list[int], target: list[int]) -> int:
        if Counter(nums)!=Counter(target):
            return 0
        start = n = len(nums)
        for i in range(n):
            if nums[i]!=target[i]:
                start = i
                break
        # find start point of suffix of target : target[i]>=target[i+1]
        ordered_suffix_start = 0
        for i in range(n-1,0,-1):
            if target[i]<=target[i-1]:
                continue
            else:
                ordered_suffix_start = i
                break
        # nums 的起点不能大于start，不能小于ordered_suffix_start,所以答案就是夹在中间的
        print(start)
        print(ordered_suffix_start)
        return max(0,min(start,n-1) - ordered_suffix_start + 1)
```

```python
nums = [1,2,3]
target = [3,2,1]
solution = Solution()
result = solution.countSuffix(nums, target)
print(result)
```
