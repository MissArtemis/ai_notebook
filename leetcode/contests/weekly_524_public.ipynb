## 使相邻元素按位异或非零的最少删除次数

```python
# 相邻异或为零等价于相等；统计连续重复对。
```

### 比赛通过提交

[提交 755461721](https://leetcode.cn/submissions/detail/755461721/)

```python
class Solution:
    def minDeletions(self, nums: list[int]) -> int:
        ans = 0
        for i in range(1, len(nums)):
            if nums[i] == nums[i-1]:
                ans += 1

        return ans

        
```

## 有向网格中的最后访问单元格

```python
# 按方向模拟；下一步越界或重复时停在当前格子。
```

### 比赛通过提交

[提交 755462730](https://leetcode.cn/submissions/detail/755462730/)

```python
class Solution:
    def lastVisitedCell(self, grid: list[list[str]]) -> list[int]:
        record = {
            "L": (0, -1),
            "R": (0, 1),
            "U": (-1, 0),
            "D": (1, 0)
        }
        visited = set()
        m, n = len(grid), len(grid[0])
        visited.add((0, 0))
        i, j = 0, 0
        while 0 <= i < m and 0 <= j < n:
            di, dj = record[grid[i][j]]
            ni, nj = i + di, j + dj
            if not (0 <= ni < m and 0 <= nj < n) or (ni, nj) in visited:
                break
            visited.add((ni, nj))
            i, j = ni, nj

        return [i, j]
```

## 覆盖目标区间的最少区间

```python
# 合并目标区间，每次选能接上且右端最远的区间；跨目标保留覆盖末端。
```

```python
from typing import List
class Solution:
    def minSelection(self, availableIntervals: List[List[int]], targetIntervals: List[List[int]]) -> int:
        sorted_target = sorted(targetIntervals, key=lambda x: x[0])
        merged_target = []
        for start, end in sorted_target:
            if merged_target and start <= merged_target[-1][1]:
                merged_target[-1][1] = max(merged_target[-1][1], end)
            else:
                merged_target.append([start, end])
        ans = 0
        print(merged_target)
        sorted_available = sorted(availableIntervals, key=lambda x: x[0])
        covered_end = float('-inf')
        i = 0
        n = len(sorted_available)
        for start, end in merged_target:
            cur_start = max(start, covered_end)
            while covered_end < end:
                next_end = covered_end
                while i < n and sorted_available[i][0] <= cur_start:
                    next_end = max(next_end, sorted_available[i][1])
                    i += 1
                if next_end < cur_start or next_end == covered_end:
                    return -1

                ans += 1
                covered_end = next_end
                cur_start = covered_end
        return ans
```
